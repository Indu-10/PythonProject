```python
print("****{msg}****".format(msg="welcome"))
```

```python
print("****{msg:20}****".format(msg="welcome"))#for padding : is important
```

```python
print("****{msg:<20}****".format(msg="welcome"))
```

```python
print("****{msg:>20}****".format(msg="welcome"))#counts 20 from the word welcome it is left alignment
```

```python
print("****{msg:^20}****".format(msg="welcome"))#centre alignment
```

```python
print("****{:^20}****".format("welcome"))#excluding msg variable and passing the value directly also works
```

formatting numbers

```python
pi=3.14159
print("The val is {}".format(pi))
```

```python
print("The val is {:.2f}".format(pi))#{}-->place holder
```

```python
num=10000
print("The val is {:,}".format(num))
```

```python
num2=101
print("The binary value is {:b}".format(num2))#o for octal,x or X for hexa values, E for scitentific notation
```
