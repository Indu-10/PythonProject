```python
name='Indu'
f1='Apple'
f2='Orange'
print(name+" Like "+f1+" and "+f2)
```

```python
text='{} like {} and {}'
print(text.format(name,f1,f2))
```

```python
print('{} like {} and {}'.format(name,f1,f2))
```

```python
print('{0} like {1} and {2}'.format(name,f1,f2))
```

```python
print('{1} like {0} and {2}'.format(name,f1,f2))#index change the value changes
```

```python
print('{name} like {f1} and {f2}'.format(name='Indu',f1='Apple',f2='Orange'))#the values need to be assigned with the same variable name
```
